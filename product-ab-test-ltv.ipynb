## 2. Проверка гипотезы в Python

Гипотеза звучит так: пользователи из Санкт-Петербурга проводят в среднем больше времени за чтением и прослушиванием книг в приложении, чем пользователи из Москвы. Попробуйте статистически это доказать, используя одностороннюю проверку гипотезы с двумя выборками:

- Нулевая гипотеза H₀: Средняя активность пользователей в часах в двух группах (Москва и Санкт-Петербург) не различается.

- Альтернативная гипотеза H₁: Средняя активность пользователей в Санкт-Петербурге больше, и это различие статистически значимо.

```python
# Выделим данные по городам
moscow_times = df_clean.loc[df_clean['city'] == 'Москва', 'hours'].values
spb_times = df_clean.loc[df_clean['city'] == 'Санкт-Петербург', 'hours'].values

# Проверяем равенство дисперсий (тест Левена)
levene_stat, levene_p = stats.levene(spb_times, moscow_times)
equal_var = levene_p > 0.05

# Односторонний t-тест: H0: μ_СПб ≤ μ_Москва, H1: μ_СПб > μ_Москва
t_stat, p_two_sided = stats.ttest_ind(spb_times, moscow_times, equal_var=equal_var)

# Корректируем p-значение для одностороннего теста
p_one_sided = p_two_sided / 2 if t_stat > 0 else 1 - (p_two_sided / 2)

print(f"t-статистика: {t_stat:.4f}")
print(f"p-значение для одностороннего теста: {p_one_sided:.4f}")

alpha = 0.05
if p_one_sided < alpha:
    print("Отвергаем H0: среднее время активности пользователей СПб статистически значимо больше, чем у пользователей Москвы.")
else:
    print("Недостаточно доказательств отвергнуть H0: различие не является статистически значимым.")
```

## 4. Проведите оценку результатов A/B-тестирования:

- Проверьте изменение конверсии подходящим статистическим тестом, учитывая все этапы проверки гипотез.

```python
from scipy.stats import norm
import math


n1, x1 = 4952, 1377
n2, x2 = 5011, 1480


p1 = x1 / n1
p2 = x2 / n2
p = (x1 + x2) / (n1 + n2)
SE = math.sqrt(p * (1 - p) * (1/n1 + 1/n2))
z = (p2 - p1) / SE


# Односторонний тест: H₁: p₂ > p₁ (B > A)
if z > 0:
    p_value = norm.cdf(-z)  # Правый хвост: вероятность получить z или больше
else:
    p_value = 1 - norm.cdf(-z)  # Если z ≤ 0, p‑value > 0.5 (нет значимости)


print(f"p1 = {p1:.4f}, p2 = {p2:.4f}")
print(f"Объединённая конверсия p = {p:.4f}")
print(f"Стандартная ошибка SE = {SE:.5f}")
print(f"z‑статистика = {z:.3f}")
print(f"p‑value (односторонний, H₁: p₂ > p₁) = {p_value:.3f}")


if p_value < 0.05:
    print("Вывод: Конверсия в группе B статистически значимо выше, чем в A (p < 0.05).")
else:
    print("Вывод: Нет оснований утверждать, что конверсия в B выше, чем в A (p ≥ 0.05).")
```
